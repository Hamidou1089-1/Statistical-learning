<div id="theme-toggle" title="Toggle Dark/Light Mode">☀</div>

<script>
  // Add the theme toggle script
  document.addEventListener('DOMContentLoaded', function() {
    // Check for saved theme preference or use preferred color scheme
    const savedTheme = localStorage.getItem('theme');
    const prefersDark = window.matchMedia && window.matchMedia('(prefers-color-scheme: dark)').matches;

    // Set initial theme
    if (savedTheme) {
      document.documentElement.setAttribute('data-theme', savedTheme);
      updateToggleButton(savedTheme);
    } else if (prefersDark) {
      document.documentElement.setAttribute('data-theme', 'dark');
      updateToggleButton('dark');
    }

    // Toggle theme function
    function toggleTheme() {
      const currentTheme = document.documentElement.getAttribute('data-theme');
      const newTheme = currentTheme === 'dark' ? 'light' : 'dark';

      document.documentElement.setAttribute('data-theme', newTheme);
      localStorage.setItem('theme', newTheme);
      updateToggleButton(newTheme);
    }

    // Update the toggle button icon
    function updateToggleButton(theme) {
      const toggleBtn = document.getElementById('theme-toggle');
      if (toggleBtn) {
        toggleBtn.innerHTML = theme === 'dark' ? '☀' : '☾';
      }
    }

    // Add event listener to toggle button
    const toggleBtn = document.getElementById('theme-toggle');
    if (toggleBtn) {
      toggleBtn.addEventListener('click', toggleTheme);
    }
  });
</script>

```python
# Let's generate the y Bernoulli this is step 1
def Bernoulli_generate(number_of_data_point: int,
                       probability:float):
    y = np.random.binomial(1, probability, number_of_data_point)
    return y


```

```python
# step 2
sigma0 = np.array([
    [0.5, 0],
    [0, 0.5]
])
sigma1 = np.array([
    [0.4, 0],
    [0, 0.4]
])

def data_sets(number_of_data_point: int,
                mean: float,
                probability: float,
                sigma0: np.ndarray,
                sigma1: np.ndarray):
    y = Bernoulli_generate(number_of_data_point, probability)
    x_data_point0 = np.random.multivariate_normal(np.zeros(sigma0.shape[0]), sigma0, number_of_data_point)
    x_data_point1 = np.random.multivariate_normal(np.array([mean, 0]), sigma1, number_of_data_point)
    return x_data_point0, x_data_point1
```

```python
d_train = data_sets(50, 1, 0.30, sigma0, sigma1)
d_test = data_sets(1000, 1, 0.30, sigma0, sigma1)
```

```python
plt.scatter(d_train[0][:,0], d_train[0][:,1], color='b')
plt.scatter(d_train[1][:,0], d_train[1][:,1], color='r')
```

```python
plt.scatter(d_test[0][:,0], d_test[0][:,1], color='b')
plt.scatter(d_test[1][:,0], d_test[1][:,1], color='r')
```
